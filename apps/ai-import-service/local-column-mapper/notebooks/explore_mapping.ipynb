# Explore Column Mapping

This notebook is designed for exploratory data analysis and visualization of the column mapping process for Excel data. It will help in understanding how the local column mapping model works and how to utilize it effectively.

```python
# Import necessary libraries
import pandas as pd
from src.mapper import ColumnMapper

# Load an example Excel file
file_path = 'path_to_your_excel_file.xlsx'
data = pd.read_excel(file_path)

# Display the first few rows of the data
data.head()
```

```python
# Initialize the column mapper
mapper = ColumnMapper()

# Perform column mapping
mapping_result = mapper.map_columns(data)

# Display the mapping result
mapping_result
```

## Visualization of Mapping Results

In this section, we will visualize the mapping results to better understand the distribution of column roles.

```python
# Visualize the mapping results
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(data=mapping_result['role'])
plt.title('Distribution of Column Roles')
plt.xlabel('Column Role')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```
